## 0. Simulate phone dataset (*)

We want to simulate data $(x,y)$ to represent cost for phone subscriptions, with: 

- $x$ - called minutes per month
- $y$ - SEK per month 

&nbsp; a) Use ```numpy.random.normal()``` to simulate a dataset with the following requirements:(*)
- set a seed to 42 (for reproducibility and reference)
- simulate 400 x-values from the r.v. $X \sim \mathcal{N}(100, 100)$ 
- take absolute value of these x-values
- simulate noise 400 noise values from r.v. $\epsilon \sim \mathcal{N(0, 50)}$ 
- Let $y = 2x+25+\epsilon$
- plot the data set 

&nbsp; b) Now we want to remove some outliers according to this assumption: (*)
- no one talks more than 300 min using this type of subscription
- no ones costs can be negative
- plot the new dataset
- also plot ground truth using the true parameters $\beta_0 = 25, \beta_1 = 2$

&nbsp; c) Insert the values into a DataFrame (*)

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# a) Simulate the dataset
np.random.seed(42)

# Simulate 400 x-values from N(100, 100) and take absolute values
x = np.abs(np.random.normal(100, 100, 400))

# Simulate 400 noise values from N(0, 50)
error = np.random.normal(0, 50, 400)

# Compute y = 2x + 25 + noise
y = 2 * x + 25 + error

# Plot the original dataset
plt.figure(figsize=(8,6))
# plt.plot((min(x), max(x)), (0, 0), linestyle="dashed", color="orange")
# plt.plot((300,300), (min(y),max(y)), linestyle="dashed", color="green")
plt.scatter(x, y, label='Data points', alpha=0.5)
plt.xlabel('Called Minutes per Month')
plt.ylabel('Cost (SEK) per Month')
plt.title('Simulated Phone Subscription Data')
plt.grid()
plt.legend()
plt.show()



# Count outliers
num_outliers_x = np.sum(x > 300)
num_outliers_y = np.sum(y < 0)
print("Number of points x > 300 min:", num_outliers_x)
print("Number of points y < 0 kr:", num_outliers_y)

# b) Remove outliers: 
#    - Remove points where minutes > 300 
#    - Remove points where cost < 0
include = (x <= 300) & (y >= 0)
x_clean = x[include]
y_clean = y[include]

print("Length of x, outliers removed:", len(x_clean))
print("Length of y, outliers removed:", len(y_clean))

# Plot the cleaned dataset along with the ground truth line y = 25 + 2x
plt.figure(figsize=(8,6))
plt.scatter(x_clean, y_clean , label='Cleaned Data', alpha=0.5)
# Generate x values for the ground truth line
x_line = np.linspace(0, 300)
y_line = 25 + 2 * x_line
plt.plot(x_line, y_line, color='red', label='Ground truth: y = 25 + 2x')
plt.xlabel('Called Minutes per Month')
plt.ylabel('Cost (SEK) per Month')
plt.title('Cleaned Phone Subscription Data with Ground Truth')
plt.legend()
plt.show()

# c) Insert the values into a DataFrame and show the first few rows
df = pd.DataFrame({'Minutes': x_clean, 'Cost': y_clean})
df.head()
````

## 1. Train|test split (*)

Before moving on with linear regression we shall first perform a train-test-split. 

&nbsp; a) Create a train-test-split function with the following call signature: (*)

```py
def train_test_split(X: pd.DataFrame, y: pd.DataFrame, train_fraction=.7: float, random_state=42: int, replace=False: bool) -> tuple
```

that returns the tuple:
```
(X_train, X_test, y_train, y_test)
```

&nbsp; b) Now use this to split up your data into a training set and test set. Check manually that the split is performed correctly. (*)


<details>

<summary>Hint</summary>

b) Check the length of each set, and check the indices of the sorted sets that they don't overlap and are not missing. Also check that they sum up to what you expect.

</details>

<br/>

<details>

<summary>Answer</summary>

Using default 0.7: 
- length of X_train: 270
- length of X_test: 116
- length of y_train: 270
- length of X_test: 116

</details>

---

````python
def train_test_split(X: pd.DataFrame,
                     y: pd.DataFrame, 
                     train_fraction=.7, 
                     random_state=42, 
                     replace=False):
    
    X_train = X * train_fraction
    y_train = y * train_fraction
    X_test = X * random_state
    y_test = y * random_state

    return (X_train, X_test, y_train, y_test)


````
